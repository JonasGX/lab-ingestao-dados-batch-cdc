# 01 — Batch com SQL direto

Consulta os primeiros 10 clientes do PostgreSQL e monta um DataFrame. Execute as células em ordem.

```python
import os
from pathlib import Path

import psycopg2
import pandas as pd
from dotenv import load_dotenv

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
load_dotenv(PROJECT_ROOT / ".env", override=False)


def env(nome):
    valor = os.environ.get(nome, "").strip()
    if not valor:
        raise ValueError(f"Variável {nome} não definida no ambiente ou no .env")
    return valor
```

```python
def conectar_postgres(usuario, senha):
    return psycopg2.connect(
        host=env("POSTGRES_HOST"),
        port=int(env("POSTGRES_PORT")),
        dbname=env("POSTGRES_DB"),
        user=usuario,
        password=senha,
        client_encoding="UTF8",
        application_name="notebook_teste",
    )
```

```python
conn = conectar_postgres(env("POSTGRES_USER"), env("POSTGRES_PASSWORD"))
try:
    with conn, conn.cursor() as cur:
        cur.execute("select * from clientes limit 10")
        resultado = cur.fetchall()
        colunas = [desc[0] for desc in cur.description]
finally:
    conn.close()

valores = [i for i in resultado]
df = pd.DataFrame(valores, columns=colunas)
```
